```python
import scanpy as sc
import anndata
from scipy import io
from scipy.sparse import coo_matrix, csr_matrix
import numpy as np
import os
import pandas as pd
import scvelo as scv
import cellrank as cr
cr.logging.print_versions()
sce = anndata.read_h5ad("/home/yyyuxi/vascular/ourdata/Tnc_vascular/SMC_ann/scvelo/ann_sce.h5ad")
```

```python
import os
import scvelo as scv
import matplotlib.pyplot as plt

ax = scv.pl.proportions(
    sce,
    groupby='subcell',
    layers=['spliced', 'unspliced'],
    show=False
)

plt.show()

plt.savefig("spliced_unspliced_proportions.pdf", format='pdf', bbox_inches='tight')
plt.close()
```
